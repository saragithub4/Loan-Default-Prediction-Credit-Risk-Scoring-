Project Name: LOAN DEFAULT PREDICTION AND CREDIT RISK SCORING

Project Overview:
This project aims to analyse the historical LendingClub loan data to identify the factors associated with loan defaults and assess borrowers' credit risk. The analysis focuses on borrower characteristics, including annual income, credit scores, debt-to-income ratio, employment length, loan amount, interest rate, and loan purpose.

Using Python, the project aims to examine historical loan performance, identify patterns associated with repayment difficulties, and develop a predictive model to estimate the likelihood of loan default.

The project will help demonstrate how data analytics can support financial institutions in understanding credit risk, evaluating lending patterns, and making more informed lending decisions.

Business Problem:
Financial institutions face the challenge of identifying borrowers who may be unable to repay their loans. Inadequate risk assessment can result in financial losses, increased default rates, and inefficient lending decisions.

Project Limitations

2.Some fields may contain substantial missing data. The treatment of missing values could affect the analysis and model performance.

4.The analysis identifies associations within historical data. It cannot establish causation or guarantee future outcomes.

5.A model trained on LendingClub data may not perform equally well for other financial institutions, borrower populations, or economic periods.

6.Predictive performance alone is insufficient for lending decisions. Any practical deployment would require appropriate fairness assessments, explainability, privacy safeguards, and compliance with applicable laws and regulations.

```python
#removing unnecessary columns
selected_columns = [
    'id',
    'loan_amnt',
    'term',
    'int_rate',
    'installment',
    'grade',
    'sub_grade',
    'emp_length',
    'home_ownership',
    'annual_inc',
    'verification_status',
    'issue_d',
    'loan_status',
    'purpose',
    'addr_state',
    'dti',
    'delinq_2yrs',
    'earliest_cr_line',
    'fico_range_low',
    'fico_range_high',
    'inq_last_6mths',
    'open_acc',
    'pub_rec',
    'revol_bal',
    'revol_util',
    'total_acc',
    'initial_list_status',
    'application_type']

# Retain only columns available in the dataset
selected_columns = [col for col in selected_columns if col in df.columns]

df = df[selected_columns].copy()

print("Shape after column selection:", df.shape)
```

```python
#Check for duplicate records
print("Total duplicate rows:", df.duplicated().sum())
```

```python
#Check duplicate loan IDs:
if 'id' in df.columns:print("Duplicate loan IDs:", df['id'].duplicated().sum())
```

```python
#Check missing values
missing_values = pd.DataFrame({'Missing_Count': df.isnull().sum(),'Missing_Percentage': (df.isnull().sum() / len(df)) * 100})

missing_values = missing_values.sort_values(by='Missing_Percentage',ascending=False)

print(missing_values)
```

```python
#Replace missing numerical values with the median
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_columns:
    if col!="loan_status":
        df[col] = df[col].fillna(df[col].median())
```

```python
#Replace missing categorical values with Unknown
categorical_columns = df.select_dtypes(include=['string', 'category']).columns

for col in categorical_columns:
        df[col] = df[col].fillna('Unknown')
```

```python
#Check missing values
missing_values = pd.DataFrame({'Missing_Count': df.isnull().sum(),'Missing_Percentage': (df.isnull().sum() / len(df)) * 100})

missing_values = missing_values.sort_values(by='Missing_Percentage',ascending=False)

print(missing_values)
```

```python
#Clean column names
df.columns = (df.columns.str.strip().str.lower().str.replace(' ', '_'))

print(df.columns.tolist())
```

```python
#clean the term column
df['term']=df['term'].str.replace('months','',case=False).str.strip()
```

```python
df['term']=pd.to_numeric(df['term'])
```

```python
df=df.rename(columns={'term':'term_in_months'})
```

```python
#Clean employment length column
if 'emp_length' in df.columns:
        df['emp_length'] = (df['emp_length'].astype(str).str.replace('< 1 year', '0', regex=False).str.replace('10+ years', '10', regex=False)
            .str.extract(r'(\d+)')[0])

df['emp_length'] = pd.to_numeric(df['emp_length'],errors='coerce')
```

```python
#Check for invalid numerical values.
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    print(col,"Negative values:",(df[col] < 0).sum())
```

```python
#clip all the negative values in dti column to 0:
df['dti']=df['dti'].clip(lower=0)
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    print(col,"Negative values:",(df[col] < 0).sum())
```

```python
#Check important financial columns:
print("Invalid loan amounts:",(df['loan_amnt'] <= 0).sum())

print("Invalid annual income:",(df['annual_inc'] < 0).sum())

print("Invalid interest rates:",(df['int_rate'] < 0).sum())
```

```python
#correcting date formats:
if 'issue_d' in df.columns:
    df['issue_d'] = pd.to_datetime(df['issue_d'],format='%b-%Y',errors='coerce')

if 'earliest_cr_line' in df.columns:df['earliest_cr_line'] = pd.to_datetime(df['earliest_cr_line'],format='%b-%Y',errors='coerce')
```

```python
#splitting loan issue date column as 2 new columns
if 'issue_d' in df.columns:
    df['issue_year'] = df['issue_d'].dt.year
    df['issue_month'] = df['issue_d'].dt.month
```

```python
#create new column credit history age
if {'issue_d', 'earliest_cr_line'}.issubset(df.columns):
    df['credit_history_years'] = ((df['issue_d'] - df['earliest_cr_line']).dt.days/365.25)
```
